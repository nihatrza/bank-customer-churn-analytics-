```python
age_bins = [17, 30, 45, 60, 100]
age_labels = ['18-30 (Gənc)', '31-45 (Orta Yaş)', '46-60 (Böyük Yaş)', '60+ (Senior)']
df['Age_Group'] = pd.cut(df['Age'], bins=age_bins, labels=age_labels)
```

```python
cs_bins = [300, 579, 669, 739, 799, 850]
cs_labels = ['Poor', 'Fair', 'Good', 'Very Good', 'Exceptional']
df['Credit_Score_Group'] = pd.cut(df['CreditScore'], bins=cs_bins, labels=cs_labels)
```

```python
df['Balance_Category'] = np.where(df['Balance'] == 0, 'Zero Balance', 'Has Balance')
```

```python
sat_map = {1: 'Very Low', 2: 'Low', 3: 'Medium', 4: 'High', 5: 'Very High'}
df['Satisfaction_Level'] = df['Satisfaction_Score'].map(sat_map)
```

```python
numerical_cols = ['CreditScore', 'Age', 'Tenure', 'Balance', 'EstimatedSalary', 'Point Earned']

print("--- Outlier Analizi (IQR Metodu) ---")
for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"{col} sütunundakı outlier sayı: {len(outliers)}")
```

```python
import pandas as pd
from sqlalchemy import create_engine, text


df = pd.read_csv('Customer-Churn-Records.csv')


engine = create_engine('postgresql://postgres:admin123@localhost:5432/churn_db')


temp_df = df[['CustomerId', 'NumOfProducts']]
temp_df.to_sql('temp_num_products', engine, if_exists='replace', index=False)


with engine.begin() as conn:
    conn.execute(text("""
        UPDATE dim_customer d
        SET numofproducts = t."NumOfProducts"
        FROM temp_num_products t
        WHERE d.customer_id = t."CustomerId";
    """))
    conn.execute(text("DROP TABLE temp_num_products;"))

print("`numofproducts` sütunu uğurla `dim_customer` cədvəlinə əlavə olundu!")
```
